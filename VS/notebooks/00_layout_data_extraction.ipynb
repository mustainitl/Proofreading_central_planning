# Checking all the pages are under same work order or not 

```python
import pdfplumber
import re
import json

pdf_path = "../input_files/BD01529397W_layout.pdf"

field_patterns = {
    "work_order": re.compile(r"(?im)^\s*(?:ITL\s+Works\s+Order|WORK\s+ORDER)\s*:\s*([^\n/]+)"),
    "reference": re.compile(r"(?im)^\s*(?:Label\s+Reference|REFERENCE)\s*:\s*([^\n:]+)"),
    "care_code": re.compile(r"(?im)^\s*CARE\s+CODE\s*:\s*([^\n]+)"),
}


def extract_field(text, pattern):
    match = pattern.search(text)
    return match.group(1).strip() if match else None


def unique_values(values):
    unique = []
    for value in values:
        if value not in unique:
            unique.append(value)
    return unique


pages = []

with pdfplumber.open(pdf_path) as pdf:
    for page_num, page in enumerate(pdf.pages, start=1):
        text = page.extract_text() or ""

        page_data = {"page": page_num}

        for field, pattern in field_patterns.items():
            page_data[field] = extract_field(text, pattern)

        pages.append(page_data)

comparison = {}

for field in field_patterns:
    values = [page[field] for page in pages]
    unique = unique_values(values)
    missing_pages = [page["page"] for page in pages if page[field] is None]

    comparison[field] = {
        "all_same": len(unique) == 1 and unique[0] is not None,
        "values": unique,
        "missing_pages": missing_pages,
    }

result = {
    "pages": pages,
    "all_pages_match": all(field["all_same"] for field in comparison.values()),
    "comparison": comparison,
}

print(json.dumps(result, indent=2, ensure_ascii=False))
```

```python
import pdfplumber
import pandas as pd
from collections import defaultdict

pdf_path = "../input_files/BD01529397W_layout.pdf"

def get_black_layout_boxes(page):
    """Find all black filled layout boxes using rect boundaries."""
    black_rects = []
    for r in page.rects:
        w = r['x1'] - r['x0']
        h = r['bottom'] - r['top']
        ns_color = r.get('non_stroking_color')
        if (w * h) > 5000 and ns_color and ns_color[0] == 0.0 and ns_color[1] == 0.0:
            black_rects.append({'x0': r['x0'], 'y0': r['top'], 'x1': r['x1'], 'y1': r['bottom'],
                                 'width_mm': round(w * 25.4/72, 1), 'height_mm': round(h * 25.4/72, 1)})
    seen, unique = set(), []
    for b in black_rects:
        key = (round(b['x0']), round(b['y0']))
        if key not in seen:
            seen.add(key)
            unique.append(b)
    return sorted(unique, key=lambda b: b['x0'])


all_layouts = []

with pdfplumber.open(pdf_path) as pdf:
    for page_num, page in enumerate(pdf.pages, 1):
        for i, box in enumerate(get_black_layout_boxes(page), 1):
            bbox = (box['x0']+1, box['y0']+1, box['x1']-1, box['y1']-1)
            words = page.crop(bbox).extract_words()
            if not words:
                continue
            
            # Group words into lines by y-position
            rows = defaultdict(list)
            for w in words:
                rows[round(w['top'] / 3) * 3].append(w['text'])
            
            lines = [' '.join(rows[y]) for y in sorted(rows.keys())]
            
            all_layouts.append({
                'page': page_num,
                'layout_num': i,
                'width_mm': box['width_mm'],
                'height_mm': box['height_mm'],
                'full_text': '\n'.join(lines)
            })

df = pd.DataFrame(all_layouts)

# Print each layout's full text block
for _, row in df.iterrows():
    print(f"\n{'='*50}")
    print(f"Page {row['page']} | Layout {row['layout_num']} | {row['width_mm']}mm x {row['height_mm']}mm")
    print('='*50)
    print(row['full_text'])
```

```python
import pdfplumber
import pandas as pd
from collections import defaultdict, Counter

def get_layout_boxes(page, min_area=5000):
    """Detect layout boxes by dominant rectangle size — works on any PDF type."""
    candidates = []
    for r in page.rects:
        w = r['x1'] - r['x0']
        h = r['bottom'] - r['top']
        if w * h > min_area:
            candidates.append({
                'x0': r['x0'], 'y0': r['top'],
                'x1': r['x1'], 'y1': r['bottom'],
                'w': w, 'h': h,
                'width_mm':  round(w * 25.4 / 72, 1),
                'height_mm': round(h * 25.4 / 72, 1)
            })
    # Deduplicate by position
    seen, unique = set(), []
    for b in candidates:
        key = (round(b['x0']), round(b['y0']))
        if key not in seen:
            seen.add(key)
            unique.append(b)
    
    # Find dominant box size (the repeating layout boxes)
    size_counts = Counter((round(b['w']), round(b['h'])) for b in unique)
    dominant = size_counts.most_common(1)[0][0]
    
    # Return only boxes matching the dominant size (±5px tolerance)
    return sorted(
        [b for b in unique if abs(b['w'] - dominant[0]) < 5 and abs(b['h'] - dominant[1]) < 5],
        key=lambda b: b['x0']
    )


def extract_layout_text(page, box, padding=1):
    """Crop to box and return all text as a single block."""
    bbox = (box['x0']+padding, box['y0']+padding, box['x1']-padding, box['y1']-padding)
    words = page.crop(bbox).extract_words()
    if not words:
        return None
    rows = defaultdict(list)
    for w in words:
        rows[round(w['top'] / 3) * 3].append(w['text'])
    return '\n'.join(' '.join(rows[y]) for y in sorted(rows.keys()))


# --- Run on any PDF ---
# pdf_path = "../input_files/BD01529397W_layout.pdf"
pdf_path = "../input_files/BD01392316W-V06-657-Layout.pdf"

all_layouts = []
with pdfplumber.open(pdf_path) as pdf:
    for page_num, page in enumerate(pdf.pages, 1):
        for i, box in enumerate(get_layout_boxes(page), 1):
            text = extract_layout_text(page, box)
            if text:
                all_layouts.append({
                    'page': page_num, 'layout_num': i,
                    'width_mm': box['width_mm'], 'height_mm': box['height_mm'],
                    'full_text': text
                })

df = pd.DataFrame(all_layouts)

for _, row in df.iterrows():
    print(f"\n{'='*50}")
    print(f"Page {row['page']} | Layout {row['layout_num']} | {row['width_mm']}mm x {row['height_mm']}mm")
    print('='*50)
    print(row['full_text'])
```
